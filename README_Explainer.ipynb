## 5. Spacing Analysis

Spacing is analyzed primarily using the **Convex Hull** of the offensive and defensive players. This measures the 'area' covered by a team.

### Convex Hull vs. Voronoi Diagrams

- **Convex Hull**: Useful for measuring the total 'spread' of a team. A larger area indicates more spacing.
- **Voronoi Diagrams**: Useful for 'Space Control' analysis. Each point on the court is assigned to the nearest player. The area of a player's Voronoi cell represents the space they 'control'.

The `spacing_analysis.py` script provides functions to calculate and plot these metrics.

```python
from scipy.spatial import ConvexHull, Voronoi, voronoi_plot_2d

def demonstrate_spacing(positions):
    # positions is a numpy array of [x, y]
    hull = ConvexHull(positions)
    vor = Voronoi(positions)
    
    fig, ax = plt.subplots(1, 2, figsize=(14, 6))
    
    # Plot Convex Hull
    ax[0].plot(positions[:,0], positions[:,1], 'ro')
    for simplex in hull.simplices:
        ax[0].plot(positions[simplex, 0], positions[simplex, 1], 'k-')
    ax[0].set_title(f"Convex Hull (Area: {hull.volume:.2f})")
    
    # Plot Voronoi
    voronoi_plot_2d(vor, ax=ax[1])
    ax[1].plot(positions[:,0], positions[:,1], 'bo')
    ax[1].set_title("Voronoi Diagram (Space Control)")
    
    plt.show()

print("Demonstration function ready.")
```

## 7. Physics-Based Space Control (Advanced)

Standard Voronoi diagrams assume all players have the same 'reach' speed. Our advanced model uses **Time-to-Reach**, which considers:
- **Current Velocity**: A player moving fast in one direction takes longer to turn.
- **Acceleration**: How quickly a player can change speed.
- **Maximum Speed**: Physical limits of the player.

This creates a more realistic 'Heatmap' of space control, where a player's 'zone' is skewed by their momentum.
